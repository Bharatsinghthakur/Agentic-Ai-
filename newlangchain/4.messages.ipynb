Messages

Messages are the fundamental unit of context for models in Langchain. They Represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM .Messages are objects that contain: 

 * Role - identifies the message type (eg. system,user)
 

* content - Represents the actual content of the messages (like text, images, audio, documents,etc)

* Metadata - Optional fields such as response information, messages IDs, and token usage

Langchain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
load_dotenv()
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:qwen/qwen3-32b")
model.invoke("Please tell what is artificial intelligence")
```

Message Prompts 

alternatively , you can pass in a list of messages to model by providing a list of messages objects.


messages types 
* system message - tells the model how to behave and provide context for interactions
* Human message - Represents user input and interaction with the model.
* ai message - Responses generated by the model including text content, tool calls and metadata
* Tool message - Represents the outputs of tool calls 

System message

A system message represent an initial set of instructions that primes the model's behavaior. you can use a system message to set the tone define the model's role and establish guidelines for responses

Human Message

A Human message represents a user input and intercations. They can contain text , images, audio , files and any other amount of mulitmodal content.

AI message 

An AI message represents the output of a model invocation.They can include multimodal data , tool calls and provider- specific metadata that you can later access.

Tool Message

for models that support tool calling , AI message can contain tool calls . tool messages are used to pass the results of a single tool execution back to the model.

```python
from langchain.messages import SystemMessage ,HumanMessage ,AIMessage 

messages = [
    SystemMessage("You are a poetry expert"),
    HumanMessage("Write a poem on Artificial intelligence")
]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("You are helpful coding assistance")

messages = [
    system_msg,
    HumanMessage("how do create a REST API?")

]
response = model.invoke(messages)
print(response.content)
```

```python
#Detailed information through system message

from langchain.messages import SystemMessage, HumanMessage

system_msg = ("""
You are a senior python developer with expertise in web frameworks.
alwyas provide code examples and explain your reasoning.
Be concise but through in your explanations.
""")

messages = [
    system_msg,
    HumanMessage("How do I create REST API?")
]

response = model.invoke(messages)
print(response.content)
```

```python
# Message MetaData 
human_msg = HumanMessage(
    content="hello",
    name="bobby", # optional identifiy different users
    id="msg_123" # Optional:unique identifier for tracing
)
```

```python
response = model.invoke(
    [human_msg]
)
response
```

```python
from langchain.messages import AIMessage , ToolMessage

#After a model makes a tool call
# here we demostrate manually creating the messages for brevity
ai_message = AIMessage(
    content = [],
    tool_calls = [{
        "name":"get_weather",
        "args":{"location":"New Delhi"},
        "id":"call_123"
}
]
)

# Execute tool and create result message
weather_result = "Sunny 40*C"
tool_message = ToolMessage(
    content = weather_result,
    tool_call_id="call_123"
)

#continue converstaion 
messages = [
    HumanMessage("whats the weather in san Francisco?"),
    ai_message, # MOdel's tool call
    tool_message # tool execution result 
]

response = model.invoke(messages)
```
