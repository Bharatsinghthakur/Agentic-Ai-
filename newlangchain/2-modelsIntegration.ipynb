STREAMING 

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can loop to process each chunk in real time:

```python
model.invoke("write me a 50 words paragraph on Human behavior for 10 year old kid playing with friends ") #require whole output to be generated then it works
```

```python
model.stream("Write me a 100 words parargraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 100 words parargraph on Artificial Intelligence"):
    print(chunk.text,end=" ",flush=True)
```

Batch

Batching a collection of independent requests to a model can significantly improves performance and reduce costs, as the processing can be done in parallel:


```python
responses = model.batch([
    "why do parrots have colorful feathers?",
    "how do airplane fly",
    "what is quantum computing?"],
    config={
        'max_concurrency':5    # limit to 5 parallel calls
    }
)

for response in responses:
    print(response)
```
