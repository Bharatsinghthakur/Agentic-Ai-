### DEEP AGENT OVERVIEW

Build agents that can plan , use subagents , and leverage file systems for complex tasks

deep agents is standalone library for buidling agents that can tackle complex, multi-step tasks. Built on langchain and inspired bu applications like Claude Code , Deep Research and manus , deep agents come with planning capablities , file systems for context management and ability to spawan subagents.

### when to use deep agents 
use deep agents when you need agents that can :
- Handle complex , multi step tasks that require planning and decompostion 
- Manage large amount of context through file system tools
- Delegates work to specialized subagents for context isloation 
- persist memory across coverstations and threads
